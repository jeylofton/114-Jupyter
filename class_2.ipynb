# Mini Challenge 2: Sum Numbers Greater Than 5
Find the sum of all numbers greater than 5.

*Examples:*
1. Given [2,7,8,3] Then the output should be: total = 15
2. Give [1,2,3] Then the output should be: Total = 0

*Criteria:*
1. Your function should receive a list and return an integer.

The collection is: numbers: [3,5,22,2,1.6,1,6,60]

```python
def sum_greater_than_5(number):
    total = 0                  
    for n in number:       
        if n > 5:              
            total += n         
    return total              

numbers = [3, 5, 22, 2, 1.6, 1, 6, 60]
print("total =", sum_greater_than_5(numbers))   # 88
```

## Mini Challenge 3: Anagram Check

Create a function that takes two strings as parameters and returns True if they are anagrams of each other. False otherwise.

    Example:
    1. Heart and Earth
    2. Scar and Cars
    3. Stars and rats
    etc.

*Criteria:*
Two strings are anagrams of each other if and only if all of the criteria below are met:

    1. They must use all the same characters
    2. They must be of exactly the same length (chars)
    3. AI and Sorted function is not allowed.

```python
def anagram(word_a, word_b):
    if len(word_a) != len(word_b):
        return False

    for letters in word_a:
        if word_a.count(letters) != word_b.count(letters):
            return False
    
    return True

print(anagram("heart", "earth"))
```
