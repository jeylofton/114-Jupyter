## Stacks

A stack is an ordered collection of elements where items are added and removed from one end, commonly referred to as the "top". The ordering principle is referred to as LIFO (Last In, First Out)

### Simplified Version (relying on built-ins functions)

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, elements):
        self.items.append(elements)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[len(self.items)-1]

    def is_empty(self):
        return self.items == []

    def __str__(self):
        return f"Stack - {self.items}"
```

```python
s1 = Stack()
s1.push("A")
s1.push("B")
s1.push("C")
print(s1.peek())
print(s1.is_empty())
print(s1)
hold = s1.pop()
print(s1)
print(hold)
```

# CustomStack (from scratch)

```python
class CustomStack:
    def __init__(self): # Empty constructor for CustomStack
        self.top = None # This class will contain a single attribute called top

    class Node: # Inner class living inside of 'CustomStack'
        def __init__(self, data): # Constructor for Node, will require a single attribute called data to create an obj
            self.data = data
            # It will also contain a second attribute called below (This one
            self.below = None # is not mandatory in the creation of the obj))

        def __str__(self): # To String helper method to convert the obj into something readable
            return f"Node - {self.data}"

    # Next Session
    def push(self, element): # Push method will help us to add elements to the stack
        if self.is_empty(): # First scenario: If it's empty I need to set the top reference to the new mode
            self.top = self.Node(element)
        else: # Second scenario: If it's not empty, then I need to do the following:
            new_node = self.Node(element) # 1. Create a new node for the data I want to store
            new_node.below = self.top # 2. Link the NEW node to the previous one (CURRENT top)
            self.top = new_node # 3. Update the top reference to the new node

    def pop(self): # Pop method will help us to remove elements from the stack
        if self.is_empty(): # First scenario: If it's empty I will raise a reference error because there is no data to retrieve
            raise ReferenceError("Stack is empty")
        else: # Second scenario: If it's not empty I need to do the following:
            value_to_remove = self.top.data # 1. Create a variable with the top value
            self.top = self.top.below # 2. Update the top reference to the previous node (the BELOW one from the current top)
            return value_to_remove # 3. Return the variable with the value of the removed node

    def is_empty(self): # If the top attribute is None then our Stack is empty
        return self.top is None

    # Assignment 3 (Pending)
    def size(self):
        # If stack's empty then will raise an error
        if self.is_empty():
            raise ReferenceError("Stack is empty")
        else:
            # Create a variable to keep the counts of the nodes
            count = 0
            # Create a variable of the current node so we can start traveling through the nodes below
            current_node = self.top
            # We will iterate if the current node is not None, once the node is None we will stop
            while not current_node is None:
                # Each iteration we will increment the counts
                count += 1
                # And we will move the node below
                current_node = current_node.below
            # Once we reach the node with the 'None' below, let's return the count variable
            return count

    def __str__(self):
        # Should return all the values from the Stack, from top to bottom, without removing them.
        values = []
        current_node = self.top
        while current_node is not None:
            values.append(current_node.data)
            current_node = current_node.below
        return f"CustomStack - {values}"

    def peek(self):
        # Should return the value from the top node, not taking it away or removing it from the stack, just displaying the top value.
        # Take edge cases in consideration.
        if self.is_empty():
            raise ReferenceError("Stack is empty")
        else:
            return self.top.data
```

```python
stack = CustomStack()
# stack.is_empty()
stack.push("A")
stack.push("B")
stack.push("C")
hold = stack.pop()
stack.pop()
stack.pop()
```
