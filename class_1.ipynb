# OOP (Objected Oriented Programming)
Python is an Object-Oriented programming, allowing us to structure code using classes and object for a better organization and reusability.

## Advantages
- Provides clear structure to programs
- Makes code easier to mantain, reuse debug
- Helps keep your code DRY (Do not Repeat Yourself)
- Allows to rebuild reusable applications with less code

## Classes and Objects
## A class definess what an object should look like and its behavior.
An object is created based on that class.

For example:

    Class -> Fruit
    Object -> Apple, Banana, Mango
    Class -> Car
    Object -> Audi, Tesla, Ford

#### The four pillars of OOP
1. Encapsulation
2. Abstraction
3. Inhertiance
4. Polymorphism

Example of a class:
    class PostListView(ListView):

```python
class Person:
    def __init__(self, name, last_name, age):
        self.name = name
        self.last_name = last_name
        self.age = age
        self.is_active = False

    def greeting(self):
        return f"Hello from {self.name}"

    def __str__(self):
        return f"Person (name = {self.name}, last_name = {self.last_name}, age = {self.age}, is_active = {self.is_active})"
```

```python
p1 = Person("Jey", "Lofton", 44)
p1.is_active = True
print(p1)
```

```python
def count_even(collection):
    total = 0
    for n in collection:
        if n % 2 == 0:
            total += 1
    return total

numbers = [3, 5, 27, 20, 6, 3.5, 60]
print("total =", count_even(numbers))
```
