# Class 4 - Queues

A quote is an ordered collection of elements where items are added and removed from *opposite* ends, commonly referred to as the "front" and "back" or "rear" of the queue. The ordering principle is referred to as FIBO (First In, First Out) or "First come, first served"

```python
class Queue:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0)
        else:
            raise IndexError("Dequeue from an empty queue")

    def peek(self):
        if not self.is_empty():
            return self.items[0]
        else:
            raise IndexError("Peek from an empty queue")

    def size(self):
        return len(self.items)

    def __str__(self):
        return f"Queue({self.items})"
```

```python
q1 = Queue()
q1.enqueue("Edwin")
q1.enqueue("Chante")
q1.enqueue("Jey")
q1.enqueue("Austin")
q1.enqueue("Luis")
print(q1)
q1.dequeue()
print(q1)
q1.dequeue()
print(q1)
q1.dequeue()
print(q1)
q1.dequeue()
print(q1)
hold = q1.dequeue()
print(q1)
print(hold)
```

### Mini Challenge / Final Report: Reverse a list using a queue
 
 """
 You're given a list" numbers = [1,2,3,4,5]
 Create a funcation and using a queue, reverse order the numbers and returned the reverse list
 """

 *Criteria:*

 1. Make sure to create a function that receives the list and reverse list
 2. Use the built-in Queue class to reverse the list

```python
def reverse_list(data):
    queue = Queue()
    for item in data:
        queue.enqueue(item)

    reversed_data = []
    while not queue.is_empty():
        reversed_data.insert(0, queue.dequeue())

    return reversed_data

# Test it
numbers = [1, 2, 3, 4, 5]
print(reverse_list(numbers))  # [5, 4, 3, 2, 1]
```
